# Graphite C-AFM analysis

Set the raw-map directory, current unit, and field of view below.


```python
from pathlib import Path
import subprocess
import sys

repo = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
raw_dir = repo / 'data' / 'raw'
results_dir = repo / 'results'
current_unit = 'A'
scan_size_nm = (2.5, 2.5)
background = 'plane'
```

```python
subprocess.run([
    sys.executable, str(repo / 'run_all.py'), str(raw_dir),
    '--results-dir', str(results_dir),
    '--current-unit', current_unit,
    '--scan-size-nm', *map(str, scan_size_nm),
    '--background', background,
], check=True)
```
